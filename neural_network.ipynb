```python
import numpy as np
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def dense(units,x,w,b):
    output = np.zeros(units)
    for i in range(units):
        output[i] = sigmoid(np.sum(np.dot(x,w[i].T)+b))
    return output
```

```python
X=np.array([[1,2],[3,4]])
```

```python
W=np.array([[1,1],[-2,-1]])
B = 5
```

```python
output = dense(2, X, W,B)
```
