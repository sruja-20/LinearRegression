```python
def getData(datapoints):
    x = np.linspace(100,1000,datapoints)
    y = np.linspace(500,1500,datapoints)
    return x,y
```

```python
x,y = getData(100)
```

```python
def compute_cost(x,y,w,b):
    m = x.shape[0]
    cost = 0
    for i in range(m):
        fx = w * x[i] + b
        y_pred = (fx - y[i] ) ** 2
        cost += y_pred
    return 1/(2*m) * cost
```

```python
compute_cost(x,y,1,0)
```

```python
compute_cost(x,y,1,1)
```

```python
def compute_weights(x,y,w,b):
    m = x.shape[0]
    df_w = 0
    df_b = 0
    for i in range(m):
        fx = w*x[i] + b
        df_wi = ( fx -y[i])*x[i]
        df_w += df_wi
        df_bi = (fx -y[i])
        df_b += df_bi
    return df_w/m,df_b/m
```

```python
def gradient_decent(x,y,w_in,b_in,learning_rate,num_iterations):
    w = w_in
    b = b_in
    for i in range(num_iterations):
        dw,db = compute_weights(x,y,w,b)
        print(f"Cost at w:{w} & b:{b} is: ",compute_cost(x,y,w,b))
        w = w - learning_rate * dw
        b = b - learning_rate * db

    return w,b
```

```python

w,b = gradient_decent(x,y,100,100,1e-11,10000)
```
