```python
import numpy as np
```

```python
def sigmoid(z):
    return 1/ (1 + np.exp(-z))
```

```python
print(sigmoid(1))
```

```python
print(sigmoid(0))
```

```python
print(sigmoid(np.array([1,2])))
```
